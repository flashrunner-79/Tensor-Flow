```python
#Finding the mean and std (but it is high)
#Formula : VAR[X] = E[X**2] - E[X]**2

import os
from torchvision import datasets,models,transforms
import torchvision
import matplotlib.pyplot as plt
import torch

data_dir= r'/content/drive/MyDrive/data'

transform_norm = transforms.Compose([
    transforms.Resize((224,224)),
    transforms.ToTensor()
])
train = datasets.ImageFolder(os.path.join(data_dir,"Train"),transform = transform_norm)
train_load = torch.utils.data.DataLoader(train,batch_size = 32,shuffle = True)

def meanstd(loader):
     channel_sum,channel_squared_sum,batch = 0, 0, 0
     for i,_ in loader:
         channel_sum +=  torch.mean(i,dim=[0,2,3])
         channel_squared_sum += torch.mean(i**2,dim=[0,2,3])
         batch += 1
     mean = channel_sum/batch
     std = (channel_squared_sum/batch - mean**2)**0.5

     return mean,std
```

```python
meanstd(train_load)
```

```python
imagenet_mean = [0.485, 0.456, 0.406]
imagenet_std = [0.229, 0.224, 0.225]

transform_data = {
    'Train': transforms.Compose([
        transforms.Resize((224,244)),
        transforms.RandomHorizontalFlip(), #some of the data will be fliped from vertical to horizontal to train the model and get the better accuracy
        transforms.ToTensor(),
        transforms.Normalize(mean = imagenet_mean, std = imagenet_std)
    ]),

    'Test' :transforms.Compose([
        transforms.Resize (256),
        transforms.CenterCrop(224), #we are doing this because to concentrate on the object(at center or some part(according to the size)).
        #(in my thought give the value for this as same as resize() value given in the 'Train' transform (before))
        transforms.ToTensor(),
        transforms.Normalize(mean = imagenet_mean,std = imagenet_std)
    ] )}
```

```python
c_path = r'/content/drive/MyDrive/data'
data = {x : datasets.ImageFolder(os.path.join(c_path,x),transform = transform_data[x])for x in ['Train','Test']}

data_loader = {x : torch.utils.data.DataLoader(data[x],shuffle = True,batch_size = 4) for x in ["Train","Test"]}
```

```python
import numpy as np

def show_image(image,title):
  #changing the tensor image into normal image using the transpose() the size of the array
  image = image.numpy().transpose((1,2,0)) #we are using numpy() to convert tensor into numpy array becuase transpose() is a numpy function(transpose() will work only in numpy array)
  #Eg : if tensor = (3,244,244)
  #     convert into = (244,244,3) <-- normal image
  #in tensor :
  #here 3 is a index of 0
  #here 244 is a index of 1
  #here 244 is a index of 2

  #using transpose((1,2,0)) <- transforming the size into (tensor index of 1(244), tensor index of 2(244), tensor index of 0 (3)) = (244,244,3)


  #we are denormalizing the image
  #normalize Formula = (original - mean)/std
  #bringing back its original form
  mean = [0.485, 0.456, 0.406]
  std = [0.229, 0.224, 0.225]
  image = (image*std)+mean

  image = np.clip(image,0,1)
  #clip() :
  #it will check the image is between 0 to 1 if not it will convert the pixel value to 0 if it is below 0 (like : -5) and it will convert into 1 if it is above zero (like : 10)
  #Eg :
  #print("Before Clipping ", image_pixels)  # [-0.3, -0.1, 0.2, 0.7, 0.9, 1.2, 1.8, 2.5]
  #print("After clipping:      ", valid_image)  # [0,    0,   0.2, 0.7, 0.9, 1,   1,   1  ]

  plt.imshow(image) #Printing the image

  if not title == None : #if the title is not none the it will print the title
      print(title)
  plt.pause(0.001) #Pause for 1 millisecond to print the image (it's not necessary just for practise)
```
